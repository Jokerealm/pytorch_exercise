```python
# 仿射变换

import imgaug
import imgaug.augmenters as iaa
from torchvision import datasets
import torch
import matplotlib.pyplot as plt
import numpy as np  


data_folder = "./FMNIST_data"
fnimist = datasets.FashionMNIST(data_folder, train=True, download=False)

tr_images = fnimist.data
tr_labels = fnimist.targets

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
plt.imshow(tr_images[0], cmap="gray")
plt.title("Original image")
plt.show()


aug_scale = iaa.Affine(scale=2)  # 旋转
plt.imshow(aug_scale.augment_image(tr_images[0].numpy()))
plt.title("augmented image")
plt.show()
```

```python
aug = iaa.Affine(translate_px=10)  # 旋转
plt.imshow(aug.augment_image(tr_images[0].numpy()))
plt.title("augmented image")
plt.show()
```
